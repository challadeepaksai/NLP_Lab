```python
from read_data import get_data
path="/home/deepakchalla/Desktop/Desktop/NLP/Lab1/train.parquet"
data=get_data(path,n=5_00_000)
```

```python
from collections import defaultdict
counts={i:defaultdict(int) for i in range(1,5)}
for sent in data:
  for i in range(1,5):
    tokens=["<s>"]*(i-1)+sent.split()+["</s>"]*(i-1)
    for j in range(len(tokens)-i+1):
      n_gram=tuple(tokens[j:j+i])
      counts[i][n_gram]+=1
```

```python
def greedy_gen(n):
  if n==1:
    return ""
  hist=tuple(["<s>"]*(n-2))
  sent=""
  while True:
    best=0
    next=""
    for key,value in counts[n].items():
      h=tuple(key[:-1])
      c=counts[n-1].get(h,0)
      prob=value/c if c>0 else 0
      if h==hist:
        if best<prob:
          best=prob
          next=key[-1]
    if next=="" or next=="</s>":
      break
    sent+=next
    sent+=' '
    hist=hist[1:]+(next,)
  return sent
```

```python
print(greedy_gen(3))
```
